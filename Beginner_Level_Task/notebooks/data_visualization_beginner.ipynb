## 1. Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
```

## 2. Load the Sample Dataset

```python
df = pd.read_csv("../data/sample_data.csv")
df
```

## 3. Matplotlib — Line Chart
A line chart is useful for showing change across an ordered sequence.

```python
x = [1, 2, 3, 4, 5]
y = [10, 15, 13, 18, 20]
plt.plot(x, y, marker="o")
plt.title("Line Chart")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.show()
```

## 4. Matplotlib — Bar Chart
A bar chart compares values across categories.

```python
students = ["A", "B", "C", "D"]
marks = [85, 92, 78, 88]
plt.bar(students, marks)
plt.title("Bar Chart - Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

## 5. Matplotlib — Histogram
A histogram shows the frequency distribution of numerical values.

```python
rng = np.random.default_rng(42)
data = rng.normal(size=100)
plt.hist(data, bins=10)
plt.title("Histogram")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

## 6. Matplotlib — Scatter Plot
A scatter plot displays individual observations using two numerical variables.

```python
x = [5, 7, 8, 7, 2, 17, 2, 9, 4, 11, 12, 9, 6]
y = [99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86]
plt.scatter(x, y)
plt.title("Scatter Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.show()
```

## 7. Matplotlib — Pie Chart
A pie chart shows how categories contribute to a whole.

```python
labels = ["A", "B", "C", "D"]
sizes = [25, 35, 20, 20]
plt.pie(sizes, labels=labels, autopct="%1.1f%%")
plt.title("Pie Chart")
plt.show()
```

## 8. Seaborn — Line Plot

```python
line_data = [2, 4, 7, 11, 16]
sns.lineplot(x=[1, 2, 3, 4, 5], y=line_data, marker="o", color="steelblue")
plt.title("Seaborn Line Plot")
plt.show()
```

## 9. Seaborn — Bar Plot

```python
sns.barplot(x=students, y=marks, color="steelblue")
plt.title("Seaborn Bar Plot")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

## 10. Seaborn — Histogram

```python
sns.histplot(data, bins=10, kde=True, color="steelblue")
plt.title("Seaborn Histogram")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

## 11. Seaborn — Scatter Plot

```python
sns.scatterplot(x=x, y=y, color="steelblue")
plt.title("Seaborn Scatter Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.show()
```

## 12. Seaborn — Box Plot
A box plot summarizes the distribution of numerical values and can help show spread and possible outliers.

```python
box_data = [[1, 2, 5, 6, 7, 9], [3, 4, 5, 7, 8, 10]]
sns.boxplot(data=box_data, color="steelblue")
plt.title("Seaborn Box Plot")
plt.xlabel("Dataset")
plt.ylabel("Values")
plt.show()
```

## 13. Seaborn — Count Plot
A count plot displays the number of observations in each category.

```python
count_data = pd.DataFrame({"Category": ["A", "B", "A", "C", "B", "A"]})
sns.countplot(data=count_data, x="Category", color="steelblue")
plt.title("Count Plot")
plt.xlabel("Category")
plt.ylabel("Count")
plt.show()
```

## 14. Matplotlib vs Seaborn

| Matplotlib | Seaborn |
|---|---|
| Lower-level plotting control | Higher-level statistical plotting |
| Highly customizable | Concise visualization functions |
| Works directly with lists/arrays | Works naturally with Pandas DataFrames |
| Useful for detailed plot customization | Useful for exploratory statistical graphics |

## Conclusion

Matplotlib and Seaborn are complementary Python visualization libraries. Practicing their basic chart types provides a foundation for exploratory data analysis and communicating patterns through visualizations.
